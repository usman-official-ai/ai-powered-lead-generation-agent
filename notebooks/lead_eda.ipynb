```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/ml_ready_dataset.csv")
print("Data loaded")
df.head()
```

```python
print(f"Total companies: {len(df)}")
```

```python
print("By country:")
print(df["Country"].value_counts())
```

```python
print("By industry:")
print(df["Industry"].value_counts())
```

```python
df['Country'].value_counts().plot(kind='bar', color='skyblue')
plt.title('Country Distribution')
plt.show()
```

```python
df['Industry'].value_counts().plot(kind='bar', color='lightgreen')
plt.title('Industry Distribution')
plt.show()
```

```python
df['Lead Quality'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Lead Quality Distribution')
plt.show()
```
